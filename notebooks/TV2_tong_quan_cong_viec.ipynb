### 5a. Demo nghiệm thu 2.4 — theo nguyên văn kế hoạch §5.3

Kế hoạch §5.3 demo: `evaluate(pipe, X, y, scheme='nested_cv', seeds=range(10))` →
`dict_keys(['bal_acc_mean', 'bal_acc_sd', 'auc_mean', 'per_fold', 'config'])` (**5 khóa**).
Thuyết minh TV2 + ADR-008 **thêm khóa `predictions`** (cần cho McNemar, ROC, phân tích lỗi TV5)
— ô dưới chạy đúng demo và đối chiếu:

````python
from src.evaluate import (SCHEMES, _baseline_pipeline, evaluate,
                          mcnemar_test, wilson_ci)
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline

pipe = _baseline_pipeline()   # pipeline tạm — ADR-008
res = evaluate(pipe, X, y=samples["class"], scheme="original_split", split=samples["split"])
keys = list(res.keys())
print("res.keys():", keys)
plan_keys = ["bal_acc_mean", "bal_acc_sd", "auc_mean", "per_fold", "config"]
print("đúng 5 khóa của kế hoạch §5.3:", all(k in keys for k in plan_keys),
      "| thêm (ADR-008):", [k for k in keys if k not in plan_keys])

row = res["per_fold"].iloc[0]
pred = res["predictions"]
n_ok = int((pred["y_true"] == pred["y_pred"]).sum())
lo, hi = wilson_ci(n_ok, len(pred))
print(f"Tập test 34 mẫu: accuracy={row['accuracy']:.3f} ({n_ok}/{len(pred)}), "
      f"Wilson 95% [{lo:.3f}, {hi:.3f}]")

maj = evaluate(Pipeline([("clf", DummyClassifier(strategy="most_frequent"))]),
               X, samples["class"], "original_split", split=samples["split"])
mc = mcnemar_test(pred["y_true"], pred["y_pred"], maj["predictions"]["y_pred"])
print(f"McNemar (LogReg 50 gen vs lớp đa số): đúng {mc['a_correct']} vs {mc['b_correct']}"
      f"/{mc['n']}, p = {mc['p_value']:.2g}")
````

````python
# 3 sơ đồ còn lại (nested_cv chạy seeds=range(3) cho nhanh — số chính thức đọc từ CSV ở mục 5b)
rows = []
for scheme in SCHEMES:
    r = evaluate(pipe, X, samples["class"], scheme,
                 seeds=range(3), split=samples["split"])
    pf = r["per_fold"]
    rows.append({"scheme": scheme, "bal_acc": round(r["bal_acc_mean"], 3),
                 "bal_acc_sd": None if pd.isna(r["bal_acc_sd"]) else round(r["bal_acc_sd"], 3),
                 "auc": round(r["auc_mean"], 3),
                 "sens": round(pf["sensitivity"].mean(), 3),
                 "spec": round(pf["specificity"].mean(), 3),
                 "n_dong_per_fold": len(pf)})
pd.DataFrame(rows)
````
